### Unexpected sharing

The variables `a` and `b` be point to the same list:

```python
a = [1,2,3]
b = a
a[0] = 42
b
```

In this case, `b` is a _copy_ of `a`:

```python
a = [1,2,3]
b = a[:]
a[0] = 42
b
```

This happens only for composite objects. If `a` has some basic type, assignment just copies the value:

```python
a = 1
b = a
a = 42
b
```

### Concatenation of strings

Python has some "syntactic sugar" for strings concatenation. Compare:

```python
['2','2','2','2']
```

with

```python
'2' '2' '2' '2'
```

and even

```python
'2''2''2''2'
```

what is an equivalent of

```python
'2'+'2'+'2'+'2'
```

May work in a more complext expressions, altough with some limitations:

```python
2*('2''2''2''2')+2*'a'
```

This is convenient to print tables with dynamically adjusted length separators:

```python
30*'-'
```

## Which data structure is more appropriate?

Compare:

```python
fruits = [["apple", 0.9], ["orange", 1.20], ["banana", 2.50]]
```

```python
desserts = ["tiramisu", 3.50, "stawberry pavlova", 3.60, "dame blanche", 7.20]
```

```python
drinks = { "coffee": 2.50, "tea": 1.75, "juice": 2.30}
```

What of if you need to transform the data from one format to another?

```python
desserts[0::2]
```

```python
desserts[1::2]
```

```python
nested_desserts = zip(desserts[0::2],desserts[1::2])
```

```python
nested_desserts
```

```python
print(nested_desserts)
```

Aha! The `zip` object acts as a virtual list, as we can see from the documentation. We will revisit this after iterators and generators, when this will become more clear:

```python
help(zip)
```

This is why we can print an element of it one at a time:

```python
for x in nested_desserts:
    print(x)
```

But next time we try this, we have already "reached the end" of the collection, so nothing happens:

```python
for x in nested_desserts:
    print(x)
```

To avoid this, we will generate a new `zip` object and convert it to a list:

```python
list(zip(desserts[0::2],desserts[1::2]))
```

Now we can compare how one representation becomes broken after applying `set`, while another - not: 

```python
set(desserts)
```

```python
set(zip(desserts[0::2],desserts[1::2]))
```
