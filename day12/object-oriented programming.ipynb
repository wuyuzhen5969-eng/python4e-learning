### Methods and Attributes in Python

**Method**
- A method is a function defined within a class.
- Different types provide different methods according to the operations they support. For example, `str` provides `upper()`, whereas `int` does not.
- A method can be accessed through an object, e.g. `text.upper`, and called using parentheses, e.g. `text.upper()`.

**Attribute**
- An attribute is a name associated with an object.
- Attributes can refer to data (e.g. `dog.name`) or methods (e.g. `dog.bark`).
- Therefore, a method accessed through an object is an attribute, but not every attribute is a method.

**OOP connection**
- Encapsulation organizes related data and behaviour within a class or object.
- This helps keep an object's state and the operations associated with it together.

**Important distinction**
- Parameter: a variable defined in a function or method definition.
- Argument: a value passed when calling a function or method.
- Parentheses `()` indicate a function or method call; they do not themselves represent the operation.

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        print(f"{self.name} says woof!")

dog = Dog("Max")
dog.name
dog.bark()
print(type(dog))
print(type(Dog))
```

##### Inheritance   
If you have some classes that are very similar and you don't want to repeat codes, then you can use superclass and subclass

```python
class pet:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def show(self):
        print(f"I am {self.name} and I am {self.age} years old")

class dog(pet):
    def speak(self):
        print('bark')
class cat(pet):
    def __init__(self, name, age, color):
        super().__init__(name, age) # inherit from parent class
        self.color = color
    def speak(self):
        print('Meow')
```

```python
d = dog('tim',2)
c = cat('bill',3, "red")
print(d.show(), d.speak(), c.show(), c.speak())
if isinstance(c, cat):
    print("True")
```

# Practice  simulate social communication process
在一个大学里，每个学生都是一个agent。有些学生已经知道一条新消息，有些还不知道。每一轮中，知道消息的学生有机会把消息传播给其他学生。
你的目标是编写一个小型 ABM，观察消息如何在学生群体中传播。

```python
import random


class Student:
    def __init__(self, name, knows_rumour = False):
        self.name = name
        self.knows_rumour = knows_rumour
        self.friends = []

    def meet(self, student):
        self.friends.append(student)
        student.friends.append(self)
        return True

    def share_rumour_r(self):
        target = random.choice(self.friends)
        i = random.random()
        if self.knows_rumour and i < 0.5:
            target.knows_rumour = True 
        print(f"It is {target.knows_rumour} that {target.name} knows rumour in {1-i} possibility")


    def share_rumour_p(self, student):
        i = random.random()
        if self.knows_rumour and student in self.friends and i < 0.5:
            student.knows_rumour = True
        print(f"It is {student.knows_rumour} that {student.name} knows rumour in {1-i} possibility")

    def share_rumour_a(self):
        for friend in self.friends[:]:
            i = random.random()
            if self.knows_rumour and i < 0.5:
                friend.knows_rumour = True
            print(f"It is {friend.knows_rumour} that {friend.name} knows rumour in {1-i} possibility")
        




```

##optimization

meet()：构建社交网络
只负责建立双向关系，不负责传播谣言。即使重复调用，也不会添加重复的朋友。



_try_share()：定义行为规则
统一处理传播者是否知道谣言、目标是否已经知道，以及随机传播是否成功。_前单下划线表示这是供类内部使用的辅助方法的约定。



三个 share_rumour 方法：定义传播策略
随机传播只选一个朋友，定向传播针对指定学生，广播传播则对所有朋友分别尝试一次。

```python
# optimization
import random


class Student:
    def __init__(self, name, knows_rumour = False):
        self.name = name
        self.knows_rumour = knows_rumour
        self.friends = []

    def meet(self, student):
        if student is self:
            return False
        if student not in self.friends:
            self.friends.append(student) #list, avoid duplicate
        if self not in student.friends:
            student.friends.append(self)
        return True

    def _try_share(self, student, p = 0.5):
        if not self.knows_rumour:
            return False #if self does not know, share is false
        if student.knows_rumour:
            return False #if student does know, share is false
        if student not in self.friends:
            return False #they are not friends, share is false
        if random.random() < p:
            student.knows_rumour = True
            print(f"{self.name} spread the rumour to {student.name}.")
            return True
        print(f"{self.name} failed to spread the rumour to {student.name}.")
        return False
    #repeated logic, callable in the below three stratege

    def share_rumour_r(self, p = 0.3): #random communication
        if not self.friends:# no firend
            return False
        target = random.choice(self.friends)
        return self._try_share(target, p)

    def share_rumour_p(self, student = random.choice, p):
        return self._try_share(student, p)

    def share_rumour_a(self, p = 0.5):
        results = []
        for friend in self.friends[:]:
            results.append(self._try_share(friend, p))
        return results # for a sequence of result,we can create a list to contain them and return the list




```

```python
alice = Student("Alice", True)
bob = Student("Bob")
charlie = Student("Charlie")
ethan = Student("Ethan")
diana = Student("Diana")

alice.meet(bob)
alice.meet(charlie)
alice.meet(ethan)
bob.meet(diana)
diana.meet(ethan)
ethan.meet(charlie)
ethan.meet(diana)

students = [alice, bob ,charlie, ethan, diana]
for round in range(1,20):
    for student in students:
        if student == bob:
            bob.share_rumour_p(diana, 0.8)
        else:
            methods = [student.share_rumour_a, student.share_rumour_r]
            random.choice(methods)()
    T_students = [student.name for student in students if student.knows_rumour]
    count = sum(student.knows_rumour for student in students if student.knows_rumour)
    print(f"in round {round}, the total number of students who know rumour is {count} and they are {T_students}")
    if len(T_students) == len(students):
            break
```
