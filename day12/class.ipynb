## Class
Good way to structure code, Less code replication, Constrain functionality, may be easier to debug. 

```python
class Classname: # capital letter at the start
    def function(self): #lowercase_
        print("haha")

```

```python
a = Classname() # call the class
a.function() # call the method
```

```python
class Greeting:
    """a program to greet with you"""
    def hello(self):
        """say hello"""
        return "hello world!"
```

```python
l = Greeting()
l.hello()
b = Greeting()
print(l)
print(b)
l == b # they are two different objects/instances of Greeting class, although the data of the two objects can be same
```

```python
class Greeting:
    """a program to greet with you"""
    sb = "you" #class attributes, sahred by every object, good for defaults and constants
    def __init__(self, name):
        self.name = name #instance/object attributes, every object has its own copy
        
    def hello(self):
        """say hello"""
        return "hello world!"
```

```python
l = Greeting("lax")
m = Greeting("maria")
print(l.name,m.name, l.sb, m.sb)
l.sb = "me" #assigning creates an instance attribute
print(l.sb, m.sb)
Greeting.sb = "they"
print(l.sb, m.sb)
l.age = 21 #give l object an "age" attribute,Still can store more data 
#Also, can delete it 
del(l.age)
m.age # but m don't have it
```
