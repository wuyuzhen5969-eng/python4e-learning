```python
class Greeting:
    """a program to greet with you"""
    def hello(self):
        """say hello"""
        return "hello world!"
```

```python
l = Greeting()
l.hello()
b = Greeting()
print(l)
print(b)
l == b # they are two different objects/instances of Greeting class, although the data of the two objects can be same
```

```python
class Greeting:
    """a program to greet with you"""
    sb = "you" #class attributes, sahred by every object, good for defaults and constants
    def __init__(self, name):
        self.name = name #instance/object attributes, every object has its own copy
        
    def hello(self):
        """say hello"""
        return "hello world!"
```

```python
l = Greeting("lax")
m = Greeting("maria")
print(l.name,m.name, l.sb, m.sb)
l.sb = "me" #assigning creates an instance attribute
print(l.sb, m.sb)
Greeting.sb = "they"
print(l.sb, m.sb)
l.age = 21 #give l object an "age" attribute,Still can store more data 
#Also, can delete it 
del(l.age)
m.age # but m don't have it
```

##### Example class with init, str and repr

```python
class Myclass:
    """A list wrapper class to demonstrate __init__, __str__ and __repr__ methods"""
    def __init__(self, info):
        self.info = info
```

```python
a = Myclass([1,4,5])
```

```python
print(a)
a
```

```python
 class Myclass:
    """A list wrapper class to demonstrate __init__, __str__ and __repr__ methods"""
    def __init__(self, info):
        self.info = info

    def __str__(self):#overwrite "print"
        return "<" + str(self.info) + ">"

    def __repr__(self):# echo, if we want to gennerate a valid python input
        return "Myclass("+ str(self.info)+")"
```

```python
a = Myclass([1,4,5])
```

```python
print(a)
a
```

```python
a.info.append(4)
a
```

```python
a.__module__#a "knows" a lot about itself. use "dir" to find out
dir(a)
```

**data attributes override method attributes with the same name**

```python
 class Myclass:
    """A list wrapper class to demonstrate __init__, __str__ and __repr__ methods"""
    def __init__(self, info):
        self.info = info

    def size(self):
       return len(self.info)
    
    def __str__(self):#overwrite "print"
        return "<" + str(self.info) + ">"

    def __repr__(self):# echo, if we want to gennerate a valid python input
        return "Myclass("+ str(self.info)+")"
```

```python
a = Myclass([2,3,4,5])
a.size()
```

```python
a.size = 1
a.size()
```
