**Tuples and Dictionaries**  
the items() method in dictionaries returns a list of (key, value) tuples.

```python
d = dict()
d['haha'] = 1
d['lala'] = 2
for (k,v) in d.items():
    print(k,v)
```

**sorting lists of tuples**  
we can take advantage of the ability to sort a list of tuples to get a sorted version of a dictionary

```python
d = {'c':1, 'b':2, 'a':8}
d.items()#a list of tuples
sorted(d.items())
```

if we could construct a list of tuples of the form(value,key) we could sort by value.

```python
d = {'c':3, 'b':2, 'a':8}
l = list()
for k,v in d.items():
    l.append((v,k))
sorted(l, reverse = True)

```

```python
l = ['i','she','he','you','i']
d = dict()
for word in l:
    d[word] = d.get(word,0)+1
d
lst = list()
for key,val in d.items(): # [(v,k) for k,v in d.items()] - list comprehension
    newtup = (val,key)
    lst.append(newtup)
lst = sorted(lst)
for val,key in lst:
    print(key,val)
```
